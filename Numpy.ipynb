```python
#11.Filter all rows where the first feature is greater than 50.
array=np.array([[25, 40, 44],
                [56, 43, 55],
                [88, 43, 68],
                [51, 23, 45]])
print(array[array[:,0]>50])
```

```python
#12.Standardize every feature column
array=np.array([[20, 31,40],
               [32, 43, 54],
               [32, 42, 52]])
mean=np.mean(array, axis=0)
std=np.std(array, axis=0)
standardize=(array-mean)/std
print(standardize)

```

```python
#13.Create a random matrix of shape (10, 4) using a fixed seed.
import numpy as np
rng=np.random.default_rng(42)
matrix=rng.random((10,4))
print(matrix)
```

```python
#14.14. Calculate X @ weights for a feature matrix and weight vector.
x=np.array([[1, 2, 3],
           [4, 5, 6]])
weights=np.array([7, 8, 9])
print("matrix multiplication:", x@weights)
```

```python
#15.Add a bias vector using broadcasting
x=np.array([[1, 2, 3],
           [4, 5, 6]])
weights=np.array([7, 8, 9])
bias=np.array([10,11])
result=x@weights
broadcasting=result+bias
print("result:",result)
print("broadcasting:",broadcasting)
```

```python
#16.Calculate MSE between y_true and y_pred.
y_true=np.array([30,40,50])
y_pred=np.array([35,45,52])
result=y_true-y_pred
square=np.square(result)
avg=np.mean(square)
print(avg)
```
