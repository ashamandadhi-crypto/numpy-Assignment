### A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each.

# Answer:
--> A * B → Performs element-by-element multiplication. Each value in A is multiplied with the value at the same position in B.
Neural-network use: Useful when applying a mask or scaling values in a layer.
--> A @ B → Performs matrix multiplication, where rows of A are multiplied with columns of B.
Neural-network use: Used when calculating a layer’s output from input data and weights.


### A4. What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise.

# Answer:
-> Broadcasting is a NumPy feature where a smaller array or value is automatically used with a larger array during element-wise operations, without manually changing its size.
-> General rule: Compare the shapes from the rightmost dimension. The dimensions are compatible if they are equal or one of them is 1. Otherwise, broadcasting is not possible.

flatten() and ravel() both convert a multi-dimensional array into a 1D array. flatten() creates a new copy, while ravel() usually gives a view of the original array.

### A6. For a 2D array, explain axis=0 and axis=1 in an aggregation function such as np.sum() or np.mean(). Illustrate with a small example.

# Answer:
axis=0 calculates the aggregation column-wise, while axis=1 calculates it row-wise.


```python
import numpy as np
array=np.array([[10,20,30],
               [10,20,30]])
print("column wise sum:", np.sum(array, axis=0))
print("column wise mean:", np.mean(array, axis=0))
print("row wise sum:", np.sum(array, axis=1))
print("row wise mean:", np.mean(array, axis=1))
```

## D3. Debug the Model Code [8 marks]
The following code contains more than one bug:

X = np.array([[1, 2],
              [2, 3],
              [3, 4]])  # shape (3, 2)
w = np.array([0.5, 1.0, 0.2])  # intended weight vector
b = 2.0
y_pred = X @ w + b
y_true = np.array([10, 20, 30])
mse = np.mean(y_true - y_pred ** 2)
Identify every bug. Explain, in terms of shapes and operator precedence, why each is a problem, and rewrite the corrected code.

Answer:

There are two bugs in the given code:

Bug 1: Incorrect shape of w

X has shape (3, 2), which means there are 3 samples and 2 features. Therefore, the weight vector must contain 2 weights. But the given w has 3 values, so X @ w is not valid.

The corrected weight vector is:

w = np.array([0.5, 1.0])

Now X has shape (3, 2) and w has shape (2,), so X @ w produces 3 predicted values.

Bug 2: Incorrect MSE expression due to operator precedence

The given expression:

mse = np.mean(y_true - y_pred ** 2)

is evaluated as y_true - (y_pred ** 2) because exponentiation ** has higher precedence than subtraction.

For MSE, we must first calculate the difference between actual and predicted values and then square it:

mse = np.mean((y_true - y_pred) ** 2)

```python

X = np.array([[1, 2],
              [2, 3],
              [3, 4]]) 
w = np.array([0.5, 1.0]) 
b = 2.0
y_pred = X @ w + b
y_true = np.array([10, 20, 30])
mse = np.mean((y_true - y_pred)** 2)
print("Predicted values:", y_pred)
print("MSE:", mse)
```

```python
#Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns

X=np.array([[10, 100], [20, 200], [30, 300]]) 
min_values=X.min(axis=0) 
max_values=X.max(axis=0) 
X_normalized = (X - min_values) / (max_values - min_values) 
print("Original matrix:") 
print(X) 
print("Normalized matrix:") 
print(X_normalized)
```
